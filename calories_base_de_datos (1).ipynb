Al no tener datos duplicados ni datos nulos, se procede a buscar  columna por columna datos atipicos. Para ello se usa una tecnica de dispersión estadistica IQR que establece un intervalo y las muestras que esten por fuera se eliminan.

**Columna "Sex"**

```python
print(f"Los géneros son",data["Sex"].unique())#No tiene datos atipicos
```

**Columna "Age"**


```python
print(f"El maximo de edad {data["Age"].max()} años \nEl minimo de edad es {data["Age"].min()} años") #Se conoce el rango de edad
```

```python
Q1=data['Age'].quantile(0.25)
Q3=data['Age'].quantile(0.75)
IQR=Q3-Q1
lim_inferior=Q1-1.5*IQR
lim_superior=Q3+1.5*IQR
data=data[(data['Age']>=lim_inferior) & (data['Age']<=lim_superior)]
data.info() #No se eliminaron datos
```

**Columna "Height"**

```python
print(f"El máximo de altura es {data["Height"].max()}")
```

```python
print(data[data['Height']>210.0])
```

```python
print("la altura minima es ", data['Height'].min())
```

```python
Q1=data['Height'].quantile(0.25)
Q3=data['Height'].quantile(0.75)
IQR=Q3-Q1
lim_inferior=Q1-1.5*IQR
lim_superior=Q3+1.5*IQR
data=data[(data['Height']>=lim_inferior) & (data['Height']<=lim_superior)]#Elimina 14 datos atipicos (alturas muy bajas o muy altas)
data.info()
```


**Columna "Weight"**



```python
print(f"El peso maximo es {data['Weight'].max()} \nEl peso minimo es {data['Weight'].min()} ")
```

```python
#Al no tener un rango dado buscamos con IQR datos atipicos
Q1=data['Weight'].quantile(0.25)
Q3=data['Weight'].quantile(0.75)
IQR=Q3-Q1
lim_inferior=Q1-1.5*IQR
lim_superior=Q3+1.5*IQR
data=data[(data['Weight']>=lim_inferior) & (data['Weight']<=lim_superior)]#Elimino 4 datos atipicos
data.info()
```

**Columna "Duration"**

```python
print(f"La duración minima es {data["Duration"].min()} y la duración máxima {data['Duration'].max()}")
```

```python
Q1=data['Duration'].quantile(0.25)
Q3=data['Duration'].quantile(0.75)
IQR=Q3-Q1
lim_inferior=Q1-1.5*IQR
lim_superior=Q3+1.5*IQR
data=data[(data['Duration']>=lim_inferior) & (data['Duration']<=lim_superior)]
data.info()#No hay datos atipicos en esta columna
```

**Columna "Heart_Rate"**

```python
print(f"El ritmo cardiaco maximo {data['Heart_Rate'].max()} y el minimo {data['Heart_Rate'].min()}")
```

```python
data[(data['Heart_Rate'] < 75) & (data['Age'] < 30)] #En personas jovenes una frecuencia cardiaca menor a 75 estando en actividad fisica es inusual
```

```python
Q1=data['Heart_Rate'].quantile(0.25)
Q3=data['Heart_Rate'].quantile(0.75)
IQR=Q3-Q1
lim_inferior=Q1-1.5*IQR
lim_superior=Q3+1.5*IQR
data=data[(data['Heart_Rate']>=lim_inferior) & (data['Heart_Rate']<=lim_superior)]
data.info() #Elimina 36 datos atipicos
```

**Columna "Body_temp"**

```python
print(f"La temperatura corporal maxima es {data['Body_Temp'].max()} y la minima {data['Body_Temp'].min()}") #Una temperatura corporal en mayor a 40,5 es inusual
```

```python
Q1_temp = data['Body_Temp'].quantile(0.25)
Q3_temp = data['Body_Temp'].quantile(0.75)
IQR_temp = Q3_temp - Q1_temp

limite_inferior_temp = Q1_temp - (1.5 * IQR_temp)
limite_superior_temp = Q3_temp + (1.5 * IQR_temp)

data=data[(data['Body_Temp'] >= limite_inferior_temp) & \
                   (data['Body_Temp'] <= limite_superior_temp)]
data.info() #Elimina 14919 datos
```

**Columna "Calories"**

```python
print("minimo de calorias quemadas es de ", data['Calories'].min(), "y el maximo es de ", data['Calories'].max())
```

```python
Q1=data['Calories'].quantile(0.25)
Q3=data['Calories'].quantile(0.75)
IQR=Q3-Q1
lim_inferior=Q1-1.5*IQR
lim_superior=Q3+1.5*IQR
data=data[(data['Calories']>=lim_inferior) & (data['Calories']<=lim_superior)]
data.info() #Se eliminaron 141 datos
```

Finalmente, se eliminaron 15114 datos del data set
