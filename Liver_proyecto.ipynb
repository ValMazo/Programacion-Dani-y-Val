```python
import pandas as pd
import numpy as np
file_path = r"/content/drive/MyDrive/Indian Liver Patient Dataset (ILPD).csv"
datos = pd.read_csv(file_path, header=None)
datos.columns = ['Age',
    'Gender',
    'Total_Bilirubin',
    'Direct_Bilirubin',
    'Alkaline_Phosphotase',
    'Alamine_Aminotransferase',
    'Aspartate_Aminotransferase',
    'Total_Proteins',
    'Albumin',
    'Albumin_and_Globulin_Ratio',
    'Dataset']
```

```python
#Elimino filas duplicados
datos = datos.drop_duplicates().reset_index(drop=True)
```

```python
#Relleno datos nulos en la columna
datos = datos.fillna(datos.mean(numeric_only=True))
```
