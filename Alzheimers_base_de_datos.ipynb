Al no tener datos nulos ni duplicados, se procede a revisar algunas columnas con el método estadístico IQR para hallar datos atípicos en las características cuantitativas, mientras que se verifica que en las categoricas no haya números distintos a los preestablecidos.

*Columna BMI*

```python
Q1=data["BMI"].quantile(0.25)#Tecnica de dispersión estadistica establece el limite superior e inferior de los datos y ayuda a eliminar datos atipicos
Q3=data["BMI"].quantile(0.75)
IQR=Q3-Q1
lim_inferior=Q1-1.5*IQR
lim_superior=Q3+1.5*IQR
data=data[(data["BMI"]>=lim_inferior) & (data["BMI"]<=lim_superior)]
data.info() #No hay datos atipicos :(
```
